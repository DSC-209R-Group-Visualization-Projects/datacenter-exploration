# Data centers
*Narrative prototype. The markdown carries the story. Lets pretend that each markdown cell is our narrative that will come from scrolling in D3 and each visualization is how the scrollytelling will change the visualization.*

```python
import sys, warnings
from pathlib import Path
ROOT = Path.cwd()
sys.path[:0] = [str(ROOT / ".pylibs"), str(ROOT / "scripts")]
warnings.filterwarnings("ignore")
import numpy as np, pandas as pd
import plotly.graph_objects as go
CLEAN = ROOT / "data" / "clean"
dc = pd.read_csv(CLEAN / "datacenters.csv")
operating = dc[dc.status_group == "operating"]   # "Operating" + "Expanding"
planned = dc[dc.status_group == "pipeline"]      # proposed, approved, permitted, under construction
```

## 1. Datacenter build-out during the AI era (roughly 2020 to now)

The United States has undergone a data center expansion in recent years. In the past five years alone, during the AI boom, the number of data centers in the US has grown from **[xx]** to **[xx]**, a **[xx]%** increase.

<span style="color:#b00">**PLACEHOLDER:** our database has no build dates, so it cannot supply these three numbers. They need an outside time series (options in the README / team notes).</span>

The sites we track show where that growth has concentrated: **606 operating data centers across 39 states**. Virginia alone has 214, more than Georgia (87) and Texas (75) combined. And the wave is not over: another **975 facilities are planned**. If they are all built, the number of tracked sites would grow by **161%**.

```python
# Numbers quoted above (recomputed so the text stays honest if the data changes)
by_state = operating.groupby("state").size().sort_values(ascending=False)
print(f"operating: {len(operating)} sites in {operating.state.nunique()} states")
print("top states:", by_state.head(3).to_dict())
print(f"planned: {len(planned)} sites -> +{len(planned) / len(operating) * 100:.0f}% if all built")
```

```python
# Map 1: current data center sites. States shaded by count of operating sites, one dot per site.
counts = by_state.rename("n").reset_index()
fig = go.Figure()
fig.add_trace(go.Choropleth(locations=counts.state, z=counts.n, locationmode="USA-states",
    colorscale=[[0, "#eef3fa"], [0.15, "#9ecae1"], [1, "#08306b"]], zmin=0, zmax=counts.n.max(),
    marker_line_color="white", marker_line_width=0.8, colorbar=dict(title="operating<br>data centers", len=0.55),
    hovertemplate="%{location}: %{z} operating data centers<extra></extra>"))
fig.add_trace(go.Scattergeo(lon=operating.lon, lat=operating.lat, mode="markers", name="operating data center",
    marker=dict(size=4.5, color="#d94801", opacity=0.85, line=dict(width=0.3, color="white")),
    text=operating.facility_name.fillna("(unnamed)") + "<br>" + operating.city.fillna("") + ", " + operating.state,
    hovertemplate="%{text}<extra></extra>"))
fig.update_layout(height=600, margin=dict(l=0, r=0, t=50, b=0), legend=dict(x=0.02, y=0.06),
    title=dict(text=f"Where today's data centers are: {len(operating)} operating sites in {operating.state.nunique()} states", x=0.02, font=dict(size=16)),
    geo=dict(scope="usa", projection=dict(type="albers usa"), showlakes=False, bgcolor="white"))
fig.write_html(ROOT / "jppolizzi" / "story_01_current_datacenters.html", include_plotlyjs=True, config={"displayModeBar": False})
fig.show()
```

*Note for the D3 version: states with no tracked site are left unshaded. The tracked database covers large and newsworthy facilities, so it undercounts small data centers, especially enterprise server rooms and smaller colocation sites.*
